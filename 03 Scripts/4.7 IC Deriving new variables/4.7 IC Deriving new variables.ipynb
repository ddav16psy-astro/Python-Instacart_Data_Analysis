# 02.01(a) If-elif-else

```python
# Define a function

def price_label(row):
    if row['prices'] <= 5:
        return 'Low range product'
    elif (row['prices'] > 5) and (row['prices'] <= 15):
        return 'Mid range product'
    elif row['prices'] > 15:
        return 'High range product'
    else: return np.nan
```

```python
# Apply the function

df['price_range'] = df.apply(price_label, axis=1)
```

```python
# Run .value_counts() function to check values frequency, use dropna = False to included any NaN value counts

df['price_range'].value_counts(dropna = False)
```

# 02.02 Use a for-loop to create a new column

# Use for-loop to create a new column classifying how busy each day of the week is (based on transaction counts)

```python
result = []

for value in ords_prods_merge["orders_day_of_week"]:
    if value == 0:
        result.append("Busiest day")
    elif value == 4:
        result.append("Least busy")
    else:
        result.append("Regularly busy")                    
```

```python
# Check first 5 elements of the created "result" list
result[:5]
```

NOTE: The for-loop syntax followed the Exercise's example of using " (double quotes) rather than ' (single quotes) but the for-loop operation still worked

```python
# Need to add the "result" list as a new column in the dataframe

ords_prods_merge['busiest_day'] = result
```

```python
# Check the frequency of the new column's values and compare with the orders_day_of_week frequencies output earlier (above)

ords_prods_merge['busiest_day'].value_counts(dropna = False)
```

# Step 2 - Add a new column using a for-loop

```python
# Add a new column that relabels transaction frequency counts to expand inclusion to the 2 busiest days and 2 slowest days
# From prior results, the 2 busiest days are Saturday (=0) and Sunday (=1), two slowest day are Tuesday (=3) and Wednesday (=4)
```

```python
outcome = []

for x in ords_prods_merge["orders_day_of_week"]:
    if x in [0, 1]:
        outcome.append("Busiest days")
    elif x in [3, 4]:
        outcome.append("Slowest days")
    else:
        outcome.append("Regularly busy")   
```

```python
# Check first 5 elements of the created "outcome" list
outcome[:5]
```

```python
# Need to add the "outcome" list as a new column labeled 'busiest_days' (plural form) in the dataframe

ords_prods_merge['busiest_days'] = outcome
```

# Step 4 - Add a new column using .loc() function

```python
# Step 4: Create a new column identifying busiest hours of day broken down into 3 periods, "Most orders", "Average orders" and "Fewest orders"
# We start by looking at the frequency distribution of the 'order_hour_of_day' column to figure out period labeling

ords_prods_merge['order_hour_of_day'].value_counts(dropna = False)
```

__Based on the above frequency distribution we label time periods as follows: "Most orders" is order_hour_of_day whose transaction count > 2000000, "Average orders" is < 2000000 and >= 1000000, and "Fewest orders" is < 1000000.  We use an loc() function-based variable derivation rather than a for-loop approach to save RAM usage. Then we output a frequency distribution.__

```python
ords_prods_merge.loc[ords_prods_merge['order_hour_of_day'].isin([9,10,11,12,13,14,15,16,17]), 'busiest_period_of_day'] = 'Most orders'
```

```python
ords_prods_merge.loc[ords_prods_merge['order_hour_of_day'].isin([8,18,19]), 'busiest_period_of_day'] = 'Average orders'
```

```python
ords_prods_merge.loc[ords_prods_merge['order_hour_of_day'].isin([0,1,2,3,4,5,6,7,20,21,22,23]), 'busiest_period_of_day'] = 'Fewest orders'
```
