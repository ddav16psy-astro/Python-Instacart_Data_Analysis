# Step 6 - Add a column followed by a flag column

```python
# Step 6: Calculate average price per customer across all their orders
# Then create a spending flag for each customer that labels price < 10 as "Low spender" and >= 10 as "High spender"
```

```python
ords_prods_merge['mean_order'] = ords_prods_merge.groupby(['user_id'])['prices'].transform('mean')
```

```python
# Check the output
ords_prods_merge[56:61]
```

```python
# Check the dimensions
ords_prods_merge.shape
```

```python
# Create the spending flag
```

```python
ords_prods_merge.loc[ords_prods_merge['mean_order'] >= 10, 'spending_flag'] = 'High spender'
```

```python
ords_prods_merge.loc[ords_prods_merge['mean_order'] < 10, 'spending_flag'] = 'Low spender'
```

```python
# Check the output
ords_prods_merge[56:61]
```

```python
# Rename 'mean_order' column (not descriptive enough) to 'mean_price_per_order'

ords_prods_merge.rename(columns = {'mean_order':'mean_price_per_order'}, inplace=True)
```

```python
# Check the output
ords_prods_merge[56:61]
```

```python
# Check the dimensions
ords_prods_merge.shape
```

# Step 7 - Add another column followed by a flag column

```python
# Step 7: Determine median per customer for 'days_since_prior_order' column, to help determine frequent vs non-frequent customers
# Create an order frequency flag based off the median, to label each customer's ordering behavior
```

# NOTE to self: I created a 1000-row ords_prods_merge dataframe subset as a "sandbox" to test the median-calculating line of code before applying it to the full dataframe. A WarningMessage alerted me to the fact that I had created a "view" (which apparently stays connected to, and if operated on, might alter the "parent" dataframe it was created from).  If then altering a view (rather than just displaying it), python gets confused, does not like that. ChatGPT offered a clean solution: when planning to alter the view (i.e. adding a column, etc.), always, always, always make a COPY of the view, then operate on the COPY.  The copy "severs" the link with the "parent" dataframe.  Correct syntax to create the copy was: df_sub = ords_prods_merge[:1000].copy(). Will adopt this copy-based approach (if planning to alter a subset view for sandbox purposes, rather than simply display the view) going forward.

```python
ords_prods_merge['median_days_since_prior_order'] = ords_prods_merge.groupby(['user_id'])['days_since_prior_order'].transform('median')
```

```python
# Check the output to compare 2 customers in the row area that transitions to the 2nd customer
# This approach saves RAM usage, memory in the current session. It's not necessary to look at 100+ rows for verification purposes.
ords_prods_merge[56:61]
```

```python
# Check the dimensions
ords_prods_merge.shape
```

```python
# Create the order frequency flag
```

```python
ords_prods_merge.loc[ords_prods_merge['median_days_since_prior_order'] > 20, 'order_frequency_flag'] = 'Non-frequent customer'
```

```python
ords_prods_merge.loc[(ords_prods_merge['median_days_since_prior_order'] > 10) & (ords_prods_merge['median_days_since_prior_order'] <= 20), 'order_frequency_flag'] = 'Regular customer'
```

```python
ords_prods_merge.loc[ords_prods_merge['median_days_since_prior_order'] <= 10, 'order_frequency_flag'] = 'Frequent customer'
```

```python
# Check the output
ords_prods_merge[56:61]
```

```python
# Check the dimensions
ords_prods_merge.shape
```

```python
# Check the distribution
ords_prods_merge['order_frequency_flag'].value_counts()
```
