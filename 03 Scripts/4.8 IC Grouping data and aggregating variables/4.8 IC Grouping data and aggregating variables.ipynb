# 03. Aggregating data with .agg()

```python
# Performing a single aggregation in 2 steps (not 3)
# The groupby() function is assigned to the df dataframe, creating a pandas Object for "department_id")
# Then the agg() function is applied onto the Object, to return the mean of the given column "order_number"

df.groupby('department_id').agg({'order_number': ['mean']})
```

```python
# Some aggregations can be done without the agg() function, as below
# This achieves the same results as the example above, with different syntax

df.groupby('department_id')['order_number'].mean()
```

```python
# In some cases, you can use dot notation, as below
# Dot notation HOWEVER is not a best practice (except for applying pandas functions, looking for dataframe attributes, etc.)
# Using it as in the example below can cause confusion to Python and/or colleagues interpreting your code -- so the brackets usage examples are better

df.groupby('department_id').order_number.mean()
```

# 04. Performing multiple aggregations

```python
# Expanding from producing a single statistic using agg(), to producing multiple statistics

df.groupby('department_id').agg({'order_number': ['mean', 'min', 'max']})
```

# 05. Aggregating data with .transform()

```python
# Looking to create a loyalty flag column after aggregating data with the transform() function
# Loyalty will be defined by the max order count size of a customer
# With transform, and calling the max() function from the NumPy library, the 3-step grouping and aggregation operations can be done in 1 line of code
# Note that 2 new columns will be created, initially a 'max_order' column with the help of transform, and later a loyalty flag column
# Do this for the for the full dataframe, not the subset used in examples above

ords_prods_merge['max_order'] = ords_prods_merge.groupby(['user_id'])['order_number'].transform('max')
```

# In response to a Python FutureWarning message on the 1-line code above, I substituted 'max' for the Exercise 4.8 use of np.max in the transform() function. After the substitution, Python did not issue a warning.

# 06. Deriving columns with .loc()

```python
ords_prods_merge.loc[ords_prods_merge['max_order'] > 40, 'loyalty_flag'] = 'Loyal customer'
```

```python
ords_prods_merge.loc[(ords_prods_merge['max_order'] <= 40) & (ords_prods_merge['max_order'] > 10), 'loyalty_flag'] = 'Regular customer'
```

```python
ords_prods_merge.loc[ords_prods_merge['max_order'] <= 10, 'loyalty_flag'] = 'New customer'
```

```python
# Output frequency distribution for the newly-created loyalty_flag column

ords_prods_merge['loyalty_flag'].value_counts(dropna = False)
```

```python
# Check a slice of columns and rows of the altered dataframe that catch the transition from one customer to the next one

ords_prods_merge[['user_id', 'loyalty_flag', 'max_order']][53:62]
```
