## 02.05 Create a data dictionary

```python
# Import a departments.csv file and create a new dataframe (which we will use to create a data dictionary)
```

```python
df_dep = pd.read_csv(os.path.join(path, 'Data', 'Original Data', 'EX 4.3', 'departments.csv'), index_col = False)
```

```python
# Examine it briefly
```

```python
df_dep.head()
```

```python
# Transposing df_dep dataframe into a new dataframe to avoid visual-only output and overwriting original dataframe
df_dep_t = df_dep.T
```

```python
# Check the result
df_dep_t
```

```python
# The "0" in the header is problematic so we will add an index to the dataframe in order to do some manipulations
# This function adds an index to a dataframe that does not have one
df_dep_t.reset_index()
```

```python
# Once we have an index column, we can create a new header for the dataframe, which involves 3 Steps
# The 3 Steps are to create a new header, then remove first row in the df, then add the new header
# The iloc[] function in pandas is used for selecting data and uses the index number of a row or column to locate the data
# Take the first row of df_dep_t for the header
new_header = df_dep_t.iloc[0]
```

```python
# Verify result
new_header
```

```python
# Create a new dataframe that only copies over rows beyond the first row
# The [1:] argument tells it to copy data from row 1 onward until end of dataframe
df_dep_t_new = df_dep_t[1:]
```

```python
# Verify
df_dep_t_new
```

```python
# Finally, set the header row as the dataframe header
# df.columns can be used to show column names or be assigned column names (as is the case here)
df_dep_t_new.columns = new_header
```

```python
# Verify result
df_dep_t_new
```

```python
# Create data dictionary from the department content dataframe
# The 'index' argument tells Python to use the numbered rows as the key values for the entries in the dictionary
data_dict = df_dep_t_new.to_dict('index')
```

```python
# Verify result
data_dict
```

```python
# Check df_prods to see how we might use the data dictionary in practice
df_prods.head()
```

```python
# As an example, use data dictionary to find what department_id of 19 for row 0 means
print(data_dict.get('19'))
```

# 03. Subsetting

```python
# Follow sequence builds the pieces of the full command that lets us subset a portion of the data of interest
# Starting with indexing code showing how Python searches for the target data
df_prods['department_id']==19
```

```python
# Now let's wrap this inside df_prods[], so that it returns only products with a department_id of 19
df_prods[df_prods['department_id']==19]
```

```python
# Now we save the results into a new df_snacks dataframe
df_snacks = df_prods[df_prods['department_id']==19]
```

```python
# Verify result
df_snacks.head()
```

```python
# Alternate way to get the same subsetting result, by using a loc function
# Note: loc function looks for column names while iloc function looks for the index number of a row or column
df_snacks_2 = df_prods.loc[df_prods['department_id']==19]
```

```python
# Verify result
df_snacks_2.head()
```

```python
# A third way of getting the same result
# The isin([19]) is a list, so the loc function is looking into a list
# That can be useful if wanting to include more values for subsetting, i.e. isin([17,18,19]) to get more departments here
df_snacks_3 = df_prods.loc[df_prods['department_id'].isin([19])]
```

```python
# Verify
df_snacks_3.head()
```

# Step 7 - Use case for the isin() method

```python
# Step 7 - Need details on products customers might use to throw a party.
# Task is to find all rows from entire dataframe that include products from these departments: alcohol, deli, beverages, meat/seafood.
# Checking data_dict created earlier, because we need department_id: alcohol is 5, beverages is 7, meat seafood is 12, deli is 20.
# We will need the isin([]) method (it's not a function) from pandas to evaluate a list of values rather than a single value.
# Comment:  besides lists, isin([]) also works for tuples, series and dataframes.
df_dinner_parties = df_prods.loc[df_prods['department_id'].isin([5, 7, 12, 20])]
```
