# 02. Combining data

```python
# Import dataframe
path = r'C:\Users\dirk8\CareerFoundry Projects\03-2025 Instacart Basket Analysis'
df_ords = pd.read_csv(os.path.join(path, 'Data', 'Prepared Data', 'orders_checked.csv'), index_col = False)
```

```python
# Check the output
df_ords.head()
```

```python
# Import another, new dataframe
df_ords_prior = pd.read_csv(os.path.join(path, 'Data', 'Original Data', 'EX 4.3', 'orders_products_prior.csv'), index_col = False)
```

```python
# Check the output
df_ords_prior.head()
```

```python
# Check dimensions
df_ords.shape
```

```python
df_ords_prior.shape
```

```python
# Merge the 2 dataframes using default "inner join" type by omitting a "how" argument
# Add "indicator = True" argument to add a merge flag column, to look at matches

df_merged_large = df_ords.merge(df_ords_prior, on = 'order_id', indicator = True)
```

```python
# Check the output
df_merged_large.head()
```

```python
# Check matches in the merge flag column values
df_merged_large['_merge'].value_counts()
```

# Caveat on the output merge flag values interpretation for joins that are not "how = 'outer'"

```python
# Exercise 4.6 points out that the counts above only showed entries with a value of "both" from the default "inner join" merge
# But that does not tell us whether we have a full match
# To check for full match, we need to check the frequency of a merge flag column using a "how = 'outer'" argument, eg an "outer join" merge
# As per Exercise 4.6, doing so would show 32434489 count of "both" and a 206209 count of "left_only".  Thus, NOT a full match
```

```python
# Check the output dimensions
df_merged_large.shape
```
