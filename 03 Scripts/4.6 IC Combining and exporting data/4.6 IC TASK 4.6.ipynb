# Step 5 - Merge dataframes

```python
# Step 5: Determine a suitable way to combine df_orders_products_combined and df_prods
# To get a full picture and help me decide, I first ran an "outer" join, omitted here for laptop RAM conservation purposes
# Resulting merge flag frequencies were: "both" 32,404,859 records, "left_only" 30,200 records, "right_only" 11 records
# The dimensions were 32,435,070 x 14
# The implied proportionally tiny 0.09% of records (30,211 / 32,435,070) lying outside the overlap zone of the 2 dfs caused me to choose "inner" join
# This will merge only those records from the 2 dfs that can find a matching 'product_id'
```

```python
df_ords_prods_merge = df_orders_products_combined.merge(df_prods, on = ['product_id'], how = 'inner', indicator = True)
```

```python
# Check the output
df_ords_prods_merge.head()
```

# Step 7 - Drop _merge flag column and export as pickle file

```python
# Step 7: Export the newly created df as ords_prods_merge in a suitable format
# First we will drop the '_merge' flag column as it served its purpose, is not needed for next steps and all its values are the same

df_ords_prods_merge = df_ords_prods_merge.drop(columns = ['_merge'])
```
