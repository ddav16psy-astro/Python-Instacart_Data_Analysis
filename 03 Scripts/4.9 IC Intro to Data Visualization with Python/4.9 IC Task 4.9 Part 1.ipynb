__Note: I will change many columns (below) with an "object" datatype to 'str' ("string") just in case these are not stored as string datatypes.__

```python
# Change some column datatypes (str e.g. string is more efficient and uses less storage than an object datatype)
# Change user_id to 'str' to match the other ords_prods_merge df datatype for this column, as it's a key identifier and we will be merging dfs later
# Modify the existing dataframe
```

```python
cust_raw['date_joined'] = pd.to_datetime(cust_raw['date_joined'])
```

# Step 5(a) - Mixed-type data

```python
# Check columns of cust_raw for mixed-type data by using a custom function
# We are replacing ".applymap()" (from the Exercises) with ".map()" (which reflects updated python syntax)

for col in cust_raw.columns.tolist():
    weird = (cust_raw[[col]].map(type) != cust_raw[[col]].iloc[0].apply(type)).any(axis = 1)
    if len (cust_raw[weird]) > 0:
        print(col)
```

__Since there were no output column names when the for-loop script (above) ran, we do not have mixed-type data in the cust_raw dataframe__

# Step 6 - Combine and export data

```python
# We start by importing the ords_prods_merge file

ords_prods_merge = pd.read_pickle(os.path.join(path, 'Data', 'Prepared Data', 'ords_prods_merge_6.pkl'))
```

__Merge the 2 dataframes using an 'inner join'__

```python
# Merge cust_raw and ords_prods_merge dfs on the 'user_id' key identifier, using an 'inner join' type, include a merge flag column

ords_prods_merge_cust = ords_prods_merge.merge(cust_raw, on = ['user_id'], how = 'inner', indicator = True)
```
